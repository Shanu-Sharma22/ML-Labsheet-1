```python
# Create a sample dataset

import pandas as pd

student_data = {
    "Name": ["Aman", "Priya", "Rahul", "Sneha", "Karan",
             "Neha", "Rohit", "Anjali"],
    
    "Age": [21, 22, 20, 23, 21, 22, 24, 20],
    
    "Marks": [85, 90, 75, 88, 80, 92, 70, 95],
    
    "Department": ["MCA", "MCA", "MBA", "MCA",
                   "MBA", "MCA", "MBA", "MCA"]
}

student_df = pd.DataFrame(student_data)

student_df.to_csv("student_data.csv", index=False)

print(student_df)
```

```python
# Q11. Load a CSV dataset using Pandas
import pandas as pd

df = pd.read_csv("student_data.csv")

print(df)
```

```python
# Q12. Display the first five records using head()
print(df.head())
```

```python
# Q14. Find total number of rows and columns
rows, columns = df.shape

print("Total Rows:", rows)
print("Total Columns:", columns)
```

```python
# Q15: Display column names

print("Column Names:")
print(df.columns.tolist())
```

```python
# Q17: Descriptive statistics

print(df.describe())
```

```python
# Q18: Dataset information

df.info()
```

```python
# Q19: Identify missing values

print(df.isnull())
```

```python
# Q20: Count missing values

missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Q21: Display unique departments

unique_departments = df["Department"].unique()

print(unique_departments)
```

```python
# Q22: Frequency of departments

department_frequency = df["Department"].value_counts()

print(department_frequency)
```

```python
# Q23: Rename Marks column

df_renamed = df.rename(columns={
    "Marks": "Score"
})

print(df_renamed)
```

```python
# Q24: Select using loc[]

selected_data = df.loc[0:3, ["Name", "Marks"]]

print(selected_data)
```

```python
# Q25: Select using iloc[]

selected_data = df.iloc[0:4, 0:3]

print(selected_data)
```

```python
# Q26: Students with marks greater than 80

high_scorers = df[df["Marks"] > 80]

print(high_scorers)
```

```python
# Q27: Sort students by marks

sorted_data = df.sort_values(
    by="Marks",
    ascending=False
)

print(sorted_data)
```

```python
# Q28: Add Result column

df["Result"] = "Pass"

print(df)
```

```python
# Q29: Delete Result column

df = df.drop("Result", axis=1)

print(df)
```

```python
# Q30: Remove duplicate records

df = df.drop_duplicates()

print(df)
```

```python
# Q31: Save modified dataset

df.to_csv(
    "modified_student_data.csv",
    index=False
)

print("Modified dataset saved successfully.")
```

```python
# Q33: Histogram of marks

import matplotlib.pyplot as plt

plt.hist(
    df["Marks"],
    bins=5
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Q34: Scatter plot of Age vs Marks

import matplotlib.pyplot as plt

plt.scatter(
    df["Age"],
    df["Marks"]
)

plt.title("Age vs Marks")
plt.xlabel("Age")
plt.ylabel("Marks")

plt.show()
```

```python
# Q35: Correlation matrix and heatmap

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical columns
numeric_data = df[["Age", "Marks"]]

# Calculate correlation
correlation_matrix = numeric_data.corr()

print("Correlation Matrix:")
print(correlation_matrix)

# Create heatmap
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```
